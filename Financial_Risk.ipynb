```python
num_cols = [
    "Income",
    "Credit Score",
    "Loan Amount",
    "Assets Value",
    "Number of Dependents",
    "Previous Defaults"
]

for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
df.columns = df.columns.str.lower().str.replace(" ","_")

int_columns = [
    "age",
    "years_at_current_job",
    "marital_status_change"
]

# Fill NaN values in the specified integer columns before converting to int
for col in int_columns:
    if df[col].isnull().any(): # Check if there are any NaN values in the column
        # Fill with median, as was done for other numerical columns
        df[col].fillna(df[col].median(), inplace=True)

df[int_columns] = df[int_columns].astype(int)
```
